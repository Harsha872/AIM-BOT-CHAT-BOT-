```python
from google import genai
from google.genai.types import Content

# Initialize the client
# Make sure you have set your API key as environment variable GEMINI_API_KEY
client = genai.Client(api_key="AIzaSyD6j7PKWB4xqFnwkFiAEl4jEgeKW1TCKWM")

def chat_with_gemini(prompt):
    try:
        # Send prompt to Gemini model
        response = client.models.generate_content(
            model="gemini-2.5-flash",  # You can also try "gemini-1.5", "gemini-2.1", etc.
            contents=prompt,
        )
        return response.text
    except Exception as e:
        return f"Error: {str(e)}"

if __name__ == "__main__":
    print("Gemini AI Chatbot is ready! Type 'quit' or 'bye' to exit.")
    while True:
        user_input = input("You: ")
        if user_input.lower() in ["quit", "exit", "bye"]:
            print("Chatbot: Goodbye!")
            break
        response = chat_with_gemini(user_input)
        print("Chatbot:", response)
```
